```python
# reverse integer
n = 12345
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print(rev)
```

```python
# COUNT VOWELS AND CONSANTS
s = "Hello World"
vowels = 0
consonants = 0
for ch in s:
    if ch.isalpha():           # only letters
        if ch.lower() in 'aeiou':   # vowel?
            vowels += 1
        else:
            consonants += 1
print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
# Find Second Largest in Array
nums = [12, 35, 1, 10, 34, 1, 35]
unique = list(set(nums))   # remove duplicates
unique.sort()              # sort ascending
print("Second Largest:", unique[-2])  # second from end
```

```python
# Sum of Digits Until Single Digit
n = 9875
while n >= 10:
    total = 0
    while n > 0:
        total += n % 10
        n = n // 10
    n = total
print(n)

# OR smarter one-liner approach:
 n = 9877
 while n >= 10:
     n = sum(int(d) for d in str(n))
 print(n)
```

```python
# Print Prime Numbers in Range
n = 20
for num in range(2, n+1):
    is_prime = True
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=' ')
```

```python
# Find All Pairs with Given Sum
nums = [1, 5, 3, 7, 9, 2, 6, 4, 1, 8]
k = 10
seen = set()
for num in nums:
    complement = k - num   # what we need to pair with num
    if complement in seen:
        pair = tuple(sorted((num, complement)))
        print(pair)
    seen.add(num)
```

```python
# Fibonacci Series (N terms)
n = 8
a, b = 0, 1
for i in range(n):
    print(a, end=' ')
    a, b = b, a+b    # magic swap!
```

```python
# Count Occurrences of Each Element
nums = [1, 2, 3, 2, 1, 4, 1]
freq = {}               # empty dictionary
for num in nums:
    freq[num] = freq.get(num, 0) + 1  # count!

for key, val in freq.items():
    print(f"{key} appears {val} times")
```

```python
# Sort Without Built-in Sort
nums = [64, 34, 25, 12, 22, 11, 90]
n = len(nums)
for i in range(n-1):
    for j in range(0, n-i-1):
        if nums[j] > nums[j+1]:
            # swap adjacent elements
            nums[j], nums[j+1] = nums[j+1], nums[j]
print(nums)
```
