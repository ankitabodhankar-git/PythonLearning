```python
# PYQ PRACTISE
# count numbers 
# Step 1: input size
n = int(input())

# Step 2: input array
arr = list(map(int, input().split()))

# Step 3: counters
positive = 0
negative = 0
zero = 0

# Step 4: traversal
for num in arr:
    
    # Step 5: check conditions
    if num > 0:
        positive += 1
    elif num < 0:
        negative += 1
    else:
        zero += 1

# Step 6: print result
print(positive, negative, zero)
```

```python
# Reverse integer
class Solution(object):
    def reverse(self, x):
        rev = 0
        sign = -1 if x < 0 else 1
        x = abs(x)

        while x:
            rev = rev * 10 + x % 10
            x //= 10

        return sign * rev
```

```python
# TWO SUM
class Solution:
    def twoSum(self, nums, target):

        for i in range(len(nums)):
            for j in range(i+1, len(nums)):
                if nums[i] + nums[j] == target:
                    return [i, j]
```
