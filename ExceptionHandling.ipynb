```python
# exception is runtime error that occurs during program execution 
print(10/0) #zerodivison error

try:  #means try this code
    print(10 / 0)
except: #if something goes wrong do this
    print("Error occurred")       
```

```python
try:
    x = int(input("Enter number:"))
    print(100/x)
except:
    print("something went wrong")
```

```python
# valueError
try:
    age = int(input("Enter age: "))
except ValueError:
    print("Please enter numbers only")
```

```python
#IndexError 
try:
    arr = [10, 20, 30]
    print(arr[5])

except IndexError:
    print("Index does not exist")
```

```python
#FileNotFoundError – File Missing
file = open("data.txt", "r")
try:
    file = open("data.txt", "r")

except FileNotFoundError:
    print("File not found")
```

```python
# KeyError – Dictionary Key Missing
try:
    data = {"name": "tnuja"}
    print(data["age"])

except KeyError:
    print("Key not found")
```

```python
# Where can exceptions happen?

User input  
File handling  
Database  
List access  
Dictionary access
```

```python
# Multiple Except Blocks - to handle different errors separately with proper messages.
try:
    x = int(input("Enter number: "))
    print(100 / x)

except ValueError:
    print("Please enter valid number")

except ZeroDivisionError:
    print("Zero not allowed")
```

```python
# example  uses in dashboard,atm,payment system
try:
    arr = [10, 20, 30]
    index = int(input("Enter index: "))
    print(arr[index])

except ValueError:
    print("Index must be number")

except IndexError:
    print("Index not available")
```

```python
try:
    x = int(input("Enter number:"))
    print(50/x)
except:
    print("this is wrong")
```

```python
# ELSE Block - runs ONLY IF
try:
    x = int(input("Enter number: "))

except ValueError:
    print("Wrong input")

else:
    print("You entered:", x)
```

```python
"" try → check
except → if fail
else → if pass""
```

```python
# example
try:
    x = int(input("Enter ur age:"))
except ValueError:
    print("wrong output")
else:
    print("age save successful")
```

```python
# finally Block - runs always used in closing files,db,logging,logout,stop transaction
#when to run -  Always, whether exception occurs or not.

try:
    print(10 / 0)

except ZeroDivisionError:
    print("Error happened")

finally:
    print("Program finished")
```
