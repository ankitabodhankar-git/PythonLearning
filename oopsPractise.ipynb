```python
# practise students marks
class Student:
    def __init__(self):
        self.__marks = 0  #private var
        
    def setMarks(self,marks):
        self.__marks = marks
    def getMarks(self):
        return self.__marks
s = Student()
s.setMarks(89)
print(s.getMarks())
```

```python
class Animal:
    def eat(self):
        print("Animal can eat")
class Dog(Animal):
    def bark(self):
        print("Dog can bark")
d = Dog()
d.eat()
d.bark()
```

```python
class Lion:
    def sound(self):
        print("Roar")   #same funct diff behaior
class Cat:
    def sound(self):
        print("Meow")
animals = [Lion(), Cat()]

for a in animals:
    a.sound()
```

```python
from abc import ABC, abstractmethod
class Appliance(ABC):
    @abstractmethod
    def turnOn(self):
        pass
class Fan(Appliance):
    def turnOn(self):
        print("Fan is ON")
class AC(Appliance):
    def turnOn(self):
        print("AC is ON")
f = Fan()
f.turnOn()
```

```python
# real life examples
# ENCAPSULATION User password ,Account balance Email, phone number,Payment details
class User:
    def __init__(self, password):
        self.__password = password   # hidden

    def check_password(self, pwd):
        return self.__password == pwd
```

```python
# INHERITANCE used when role can student - user
class User:
    def login(self):
        print("User logged in")

class Admin(User):
    def delete_user(self):
        print("User deleted")
```

```python
# POLYMORPHISM used for payment, otp,,.
class UPI:
    def pay(self):
        print("Pay using UPI")
class Card:
    def pay(self):
        print("Pay using Card")
def make_payment(method):
    method.pay()
```
