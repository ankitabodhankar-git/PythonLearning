```python
# Searching - finding an ele inside of list,array
arr = [10,40,50,30,60] #used in real life find student or item in list onee by one
target = 30
for num in arr:
    if num == target:
        print("Found")
```

```python
# return index of element   also called LINEAR SEARCH
arr = [10,20,30,40,45]
target = 30
for i in range(len(arr)):
    if arr[i] == target:
        print("found at index:", i)
    
```

```python
# Moves zero to end
arr = [0,1,0,3,12]

result = []
zero_count = 0

for num in arr:
    if num == 0:
        zero_count += 1
    else:
        result.append(num)

for i in range(zero_count):
    result.append(0)

print(result)
```

```python
# REVERSE ARRAY  - swap 1st and last elemet
arr = [1,2,3,4]
left = 0
right = len(arr) -1 
while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1
print(arr)
    
```

```python
# 2nd largest
arr = [5,1,8,3]

largest = arr[0]
second = -1

for num in arr:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest:
        second = num
print(second)
```

```python
# Multiply digits    IMP 
n = 5244
product = 1

while n > 0:
    digit = n % 10
    product *= digit
    n = n // 10
print(product)
```

```python
# pallindrom  - reverse no. and compare
n = 121
temp = n
rev = 0

while n > 0:
    digit = n % 10
    rev = rev*10 + digit
    n = n // 10

print(temp == rev)
```
