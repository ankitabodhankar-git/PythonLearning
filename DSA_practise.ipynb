```python
# PRACTISE QUESTION 
# print all SUM OF 1 to N   // think in eng. then implement into code
n = int(input())
sum = 0    #start from 0
for i in range(1, n+1):     # without loop use this {Sum = n(n+1)/2}
    print(i, end=" ")  #use for print no. one by one line next line
    sum = sum +i
print("\nSum =", sum)
    
```

```python
# Sum of Digits

n = int(input())
sum = 0

while n > 0:
    digit = n % 10 #add last
    sum = sum + digit
    n = n // 10  #remove last digit
print(sum)
```

```python
# Right Triangle Patterni0:

for i in range(1,6):
    for j in range(i):
        print('*' , end=" ")
print()  # Output: # * # ** # *** # **** # *****
    
```

```python
#  Inverted Triangle
for i in range(5, 0, -1):
    for j in range(i):
        print('*', end='')
    print()
```
