```python
# Instance methods- are methods that work on object data (instance variables) and use self to access that data 
class Student:
    def __init__(self, name):
        self.name = name   # instance variable

    def display(self):     # instance method
        print(self.name)
s1 = Student("Ankita") #obj.creation
s2 = Student("Aditya")
s1.display()
s2.display()
```

```python
# Encapsulation is the process of wrapping data and methods together into a single unit (class).
class Account: it improve code and security thas why use..
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Updated balance:", self.balance)
        
```

```python
# Inheritance is a mechanism where one class acquires properties and methods of another class.
class Person:
    def speak(self):
        print("I can speak")
class Student(Person):
    def study(self):
        print("I study")
s = Student()
s.speak()   # inherited
s.study()   # own method
```

```python
# Polymorphism means same method name but different behavior same action diff result inccrese flexibility
class Dog:
    def sound(self):
        print("Bark")
class Cat:
    def sound(self):
        print("Meow")
animals = [Dog(), Cat()]
for animal in animals:
    animal.sound()
```

```python
# Abstraction means hiding internal details and showing only essential features
class Car:
    def start(self):
        print("Engine started")
```

```python
#practise question
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)
        
# Creating multiple objects
s1 = Student("Ankita", 85)
s2 = Student("Aditya", 78)
s3 = Student("Archana", 90)

s1.display()
s2.display()
s3.display()
```

```python
# Q2
class BankAccount:
    def __init__(self, balance):
        self.balance = balance   # data inside class (encapsulated)

    def deposit(self, amount):  # instance method
        self.balance = self.balance + amount
        print("Amount Deposited:", amount)

    def show_balance(self):     # instance method
        print("Current Balance:", self.balance)
acc1 = BankAccount(1000)
# Using methods
acc1.deposit(500)
acc1.show_balance()
```
