```python
# find maximum
arr = [4,7,5,9,5]
max_value = arr[0]#strt from 0 index
for num in arr:
    if num > max_value: #compare one by one ele until found big no.
        max_value = num
print(max_value)
```

```python
# find minimum
arr = [4,7,5,9,5]
min_value = arr[0]
for num in arr:
    if num < min_value:
        min_value = num
print(min_value)
```
