```python
# MODULE - is a python file (.py) that contains reusable code like functions, variables, classes
# Modules help in code reusability and maintainability

# eg1. calculator.py - this is a module
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

# main.py file
import calculator
print(calculator.add(10, 20))
print(calculator.sub(10, 20))
```

```python
# eg2. from module import function
from calculator import add
print(add(2, 3))

# eg3. import with alias (alias means short form of name)
import calculator as calc
calc.add(4, 5)
```

```python
# Types
# 1. Built-in modules are modules that are already provided by Python and can be used directly
# Examples: math, random, os, sys, datetime

import math
print(math.sqrt(16))
print(math.pow(16, 4))
print(math.factorial(46))
print(math.ceil(x))     # smallest integer greater than or equal to x
print(math.floor(x))    # largest integer less than or equal to x
print(math.sin(x))      # sine of x (x in radians)

import random
print(random.randint(1, 10))     # random integer between 1 and 10
print(random.random(1, 10))      # random float 
print(random.choice(1, 10))      # randomly choose from a sequence 

import datetime
print(datetime.datetime.now())  # current date and time
print(datetime.datetime.date()) 
print(datetime.datetime.time()) 

# Operating system (os) module — used to create folders, delete files, etc.
import os
print(os.getcwd())  # get current working directory

# System (sys) module — used for system-specific info, exit program, command-line arguments, etc.
import sys
print(sys.version)  # Python version info
```

```python
# FILE MODULES
# os.path module used for file chking existence. get file size
import os
print(os.path.exists("data.txt"))

# DIRECTORY HANDLING - Data handling modules are Python built-in modules used to store, read, write, exchange, and manage data in different formats such as files, CSV, JSON, etc.
# CSV module used to read and write data in CSV (Comma Separated Values) format also used for excel,data reports.,,
import csv
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["RollNo", "Name", "Marks"])
    writer.writerow([1, "Anuja", 85])

import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# json module used to store and exchange data in JSON format, which is lightweight and human-readable also ued for api, web application

import json
data = {"name": "Anuja", "marks": 90}
with open("data.json", "w") as file:
    json.dump(data, file)

import json
with open("data.json", "r") as file:
    data = json.load(file)
    print(data)
```

```python
# sqlite3 MODULE used to work with SQLite database, which is a lightweight database stored in a file.
import sqlite3
conn = sqlite3.connect("test.db")
conn.execute("CREATE TABLE IF NOT EXISTS student(id INT, name TEXT)")
conn.close()
```
